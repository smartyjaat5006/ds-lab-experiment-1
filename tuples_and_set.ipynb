#tuple
### ordered collection of items
## symbol()
## tuple immutable
## supports index

```python
# create a tuple
students =(101,"chirag","CSE",8.9)
print(students)

# access by index
print("Students department", students[2])

#access the negative index
print("Accessing by negative index: ",students[-2])
```

```python
subjects=("DSV","Machine Learning ","OS","Cloud","SQL","Statistics","Python")
print(subjects)

# find the length of the tuple
print(len(subjects))
```

```python
# tuple slicing

print(subjects[3:])
print(subjects[1:5])
```

```python
# tuple is mutable or immutable
print(students)
students[1]="Manish"
prints(students)
```

```python
# nested tuple
student_info = (10025,"Rahul",(89,98,78,75))
print(student_info)
print("nested tuple displaying 3rd subject mark:" ,student_info[2][2])
```

```python
position=student_info.index("Rahul ")
print("Data rahul present in the index",position)
```

# set
## unordered collection of items
## mutable
## symbol{}
## no duplication
## it will give the unique values
## union,intersection,difference, symmetric difference,superset




```python
studentsname = {"amit","sanjiv","kalai","amit","kiran","neha"}
print(studentsname)
```

```python
print(studentsname[2])
```

```python
studentsname.add("priya")
print(studentsname)

#update the set
studentsname.update(["anil","nandini","balaji"])
print(studentsname)
```

```python
#clear a data
numbers={89,78,68,97,69}
numbers.clear()
print(numbers)

del numbers
print(numbers)
```

```python
# find the element in the set
print(studentsname)

#find the element anil
print( "anil" in studentsname)
```

```python
# set operation
# union
# intersection
# difference
# symmetric difference

python_students={"rahul","priya","hari","ganesh"}
java_students={"rahul","hari","kiran","ravi","giri"}

res_union = python_students.union(java_students)
print(res_union)
```

```python
res_intersection = python_students.intersection(java_students)
print(res_intersection)
```

```python
res_difference = python_students.difference(java_students)
print(res_difference)
```
