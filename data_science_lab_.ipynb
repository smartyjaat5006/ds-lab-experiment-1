```python
# access the element using index
name =["aneesh","chirag","devi","ram"]
print(name)
print(name[-2])
```

```python
#example for mutable in the list
cgpa=[8,9,4,9.6,9.9]
print (cgpa)
cgpa[2]=7.7
print("after updation ",cgpa)
```

```python
#add an item in the list
Name=["kiran","anil","lalit","chirag"]
Name.append("deepika")
print(Name)
```

```python
#insert  in particular index
student_name =["arav","kapil","sanjeev"]
student_name.insert(1,"hariharan")
print(student_name)
```

```python
#extend function
sec_a=["chirag","jhansi","devi"]
sec_b=["kiran","ganesh","lakshman"]
sec_a.extend(sec_b)
print(sec_a)
print(sec_b)
```

```python
#remove an item in the list
sec_a_student=["chirag","jhansi","devi","anil"]
sec_a_student.remove("devi")
print(sec_a_student)
del sec_a_student[1]
print(sec_a_student)
sec_a_student.pop()
print(sec_a_student)
```

```python
#list slicing
subject=["python",
         "statistics",
         "SQL",
         "Java",
         "os",
         "machine learning"]
print(subject[1:4])
print(subject[:3])
print(subject[4:])
print(subject[::2])
#length of list
print("length of the list;",len(subject))
```

```python
marks=[78,67,89,64,57,42,51]
marks.sort()
print(marks)

#sort in decreasing order
marks.sort(reverse=True)
print(marks)
```

```python
#nested list
marks = [
    [58,61,87],
    [78,88,74],
    [56,88,78]
]
print(marks)
print(marks[1])
print(marks[2][1])
```
